```python
from transformers import pipeline

# Load the classification pipeline with the specified model
pipe = pipeline("text-classification", model="tabularisai/multilingual-sentiment-analysis")
```

```python
# Classify a new sentence
sentence = """i am from java Batch-41 Jaganath Sir is an outstanding teacher for Core Java.
His way of explaining concepts is clear, practical, and easy to understand. 
He makes even complex topics simple with real-life examples. 
His dedication and patience help students build strong programming skills. 
Highly recommended for anyone looking to learn Core Java."""


# Print the result
result = pipe(sentence)
print(result)
```
